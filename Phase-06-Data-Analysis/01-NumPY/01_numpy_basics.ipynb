```python
sales = np.array([120, 450, 80, 700, 250, 900])

# Boolean filtering
print(sales[sales > 300])
print(sales[sales < 200])

# Multiple conditions
print(sales[(sales >= 200) & (sales <= 700)])
print(sales[(sales < 100) | (sales > 800)])

# Sorting and unique values
print(np.sort(sales))
print(np.sort(sales)[::-1])
print(np.unique(np.array([1, 2, 2, 3, 3, 3])))
```

```python
import numpy as np

# Sample retail data
product_ids = np.array([101, 102, 103, 104, 105])
prices = np.array([250.0, 500.0, 150.0, 800.0, 350.0])
quantities = np.array([4, 2, 10, 1, 3])

# Calculate revenue per product
revenue = prices * quantities

# Calculate overall metrics
total_revenue = np.sum(revenue)
average_order_quantity = np.mean(quantities)
best_product_index = np.argmax(revenue)

print("Revenue per product:", revenue)
print("Total revenue:", total_revenue)
print("Average quantity:", average_order_quantity)
print("Top product ID:", product_ids[best_product_index])

# Filter high-revenue products
high_revenue_mask = revenue >= 1000

print("High-revenue product IDs:", product_ids[high_revenue_mask])
print("High-revenue values:", revenue[high_revenue_mask])

# Revenue contribution in percentage
revenue_share = revenue / total_revenue * 100
print("Revenue share (%):", np.round(revenue_share, 2))

# Validate the data
print("Any negative prices?", np.any(prices < 0))
print("Any negative quantities?", np.any(quantities < 0))
print("All prices finite?", np.all(np.isfinite(prices)))
```
